```python
##Demonstrate NumPy array creation, indexing, slicing, reshaping, broadcasting and mathematical operations on multidimensional arrays. 
import numpy as np
arr=np.array([[4,3,2],[10,1,0],[5,8,24]])
print("original array:\n",arr)
```

```python
print("indexing:\n")
print("element at[0,1]=",arr[0,1])
print("element at[-1,-1]=",arr[-1,-1])
```

```python
print("slicing:\n")
print("row:\n",arr[:1,:])
print("column:\n",arr[:,2:])
```

```python
a=np.arange(1,13)
reshaped=a.reshape(3,4)
print("reshaped array:\n")
print(reshaped)
```

```python
b=np.array([20,30,40])
print("broadcasting:\n",arr+b)
```

```python
A=np.array([[1,2,1],[2,2,3]])
B=np.array([3,4,5])
print("array A:\n",A)
print("array B:\n",B)
```

```python
add=np.add(A,B)
print("addition:\n",add)
sub=np.subtract(B,A)
print("subtraction:\n",sub)
multi=np.multiply(A,B)
print("multiplication:\n",multi)
div=np.divide(A,B)
print("dicision:\n",div)
power=np.power(A,B)
print("power:\n",power)
```
