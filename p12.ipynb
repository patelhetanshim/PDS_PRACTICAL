```python
##Use SciPy's stats module to compute statistical measures and perform statistical tests on datasets. 
import numpy as np
from scipy import stats
n = int(input("Enter number of data values: "))
data = []
for i in range(n):
    value = float(input(f"Enter value {i+1}: "))
    data.append(value)
data = np.array(data)
print("\nStatistical Measures")
print("Dataset:", data)
print("Mean:", stats.tmean(data))
print("Median:", np.median(data))
print("Mode:", stats.mode(data, keepdims=True).mode[0])
print("Variance:", stats.tvar(data))
print("Standard Deviation:", stats.tstd(data))
test_mean = float(input("\nEnter population mean for t-test: "))
t_stat, p_value = stats.ttest_1samp(data, test_mean)
print("\nStatistical Test")
print("T-statistic:", t_stat)
print("P-value:", p_value)
if p_value < 0.05:
    print("Result: Significant difference")
else:
    print("Result: No significant difference")
```
