```python
##Compute and visualize Binomial and Poisson probability distributions using SciPy and Matplotlib. 
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import binom, poisson
# ---------------------------------------
# Take dynamic input from user
# ---------------------------------------
n = int(input("Enter number of trials (n): "))
p = float(input("Enter probability of success (p, 0 to 1): "))
lam = float(input("Enter Poisson average rate (lambda): "))
# ---------------------------------------
# Validate input
# ---------------------------------------
if n <= 0:
    raise ValueError("n must be greater than 0")
if p < 0 or p > 1:
    raise ValueError("p must be between 0 and 1")
if lam <= 0:
    raise ValueError("lambda must be greater than 0")
# ---------------------------------------
# Binomial Distribution
# ---------------------------------------
x_binomial = np.arange(0, n + 1)
binomial_probability = binom.pmf(
    x_binomial,
    n,
    p
)
# ---------------------------------------
# Poisson Distribution
# ---------------------------------------
# Automatically choose suitable range
max_poisson = int(lam + 4 * np.sqrt(lam) + 5)
x_poisson = np.arange(0, max_poisson + 1)
poisson_probability = poisson.pmf(
    x_poisson,
    lam
)
# ---------------------------------------
# Display some probability values
# ---------------------------------------
print("\n Binomial Distribution")
for x, probability in zip(x_binomial, binomial_probability):
    print(f"P(X = {x}) = {probability:.4f}")
print("\n Poisson Distribution")
for x, probability in zip(x_poisson, poisson_probability):
    print(f"P(X = {x}) = {probability:.4f}")
# ---------------------------------------
# Calculate Mean and Variance
# ---------------------------------------
binomial_mean = n * p
binomial_variance = n * p * (1 - p)
poisson_mean = lam
poisson_variance = lam
print("\nStatistics")
print(f"Binomial Mean     = {binomial_mean:.4f}")
print(f"Binomial Variance = {binomial_variance:.4f}")
print(f"Poisson Mean      = {poisson_mean:.4f}")
print(f"Poisson Variance  = {poisson_variance:.4f}")
# ---------------------------------------
# Visualization
# ---------------------------------------
plt.figure(figsize=(12, 6))
# Binomial
plt.stem(
    x_binomial,
    binomial_probability,
    linefmt='b-',
    markerfmt='bo',
    basefmt=' ',
    label='Binomial'
)
# Poisson
plt.stem(
    x_poisson,
    poisson_probability,
    linefmt='r-',
    markerfmt='ro',
    basefmt=' ',
    label='Poisson'
)
# ---------------------------------------
# Graph settings
# ---------------------------------------
plt.title("Binomial and Poisson Probability Distributions")
plt.xlabel("Number of Successes / Events")
plt.ylabel("Probability")
plt.grid(True, alpha=0.3)
plt.legend()
plt.tight_layout()
plt.show()
```
