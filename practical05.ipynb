```python
##Perform Read, Write, and Append operations on text files, CSV files and Binary files using Python.
with open("example.txt", "w") as file:
    file.write("Hello World !\n")
print("Data written successfully!")
```

```python
with open("example.txt", "a") as file:
    file.write("Add New Line\n")
print("Data Added Successfully!")
```

```python
with open("example.txt", "r") as file:
    content = file.read()
    print(content)
```

```python
import csv

data = [
    ["Name", "Roll No", "Marks"],
    ["Hetanshi", "140", "99.8"],
    ["Bhumi", "151", "87.3"],
    ["Monil", "55", "97.9"],
]
```

```python
with open("Student.csv", mode="w", newline="") as file:
    writer = csv.writer(file)
    writer.writerows(data)       
print("Data written in CSV file successfully!")
```

```python
data = [
    ["Mayur", "102", "99.7"]
]
with open("Student.csv", mode="a", newline="") as file:
    writer = csv.writer(file)
    writer.writerows(data)
```

```python
with open("Student.csv", mode="r") as file:
    csvfile = csv.reader(file)
    for line in csvfile:
        print(line)
```

```python
import pickle

data = {"Name": "Hetanshi", "Roll No": 140, "Branch": "C.E."}

with open("data.bin", "wb") as file:
    pickle.dump(data, file)
print("Data written successfully.")
```

```python
add = ["record", "student"]

with open("data.bin", "ab") as file:
    pickle.dump(add, file)
```

```python
with open("data.bin", "rb") as file:
    read = pickle.load(file)
    print(read)
```
