```python
##Connect Python with MySQL database and perform Create, Read, Update and Delete (CRUD) operations. 
import mysql.connector
database = mysql.connector.connect(
    host="localhost",
    user="root",
    passwd="",
    database="ssasit"
)
cursorobject = database.cursor()
while True:
    print("\nsql Opretions")
    print("1. Create The Table In Database ssasit")
    print("2. Insert The Data In Table")
    print("3. Dispaly")
    print("4. Update In Table")
    print("5. Delete From Table")
    print("6. Exit")
    
    choice = int(input("Enter The Choice:"))
    
    if choice == 1:
        studentrecord = """CREATE TABLE IF NOT EXISTS Student
                    (Id int, Name varchar(30), Branch varchar(60))"""
        cursorobject.execute(studentrecord)
        print("Table Is Created Succesfully!!!")
        
    elif choice == 2:
        Id = int(input("Enter The Id:"))
        Name = input("Enter The Name:")
        Branch = input("Enter The Branch:")
        sql = "INSERT INTO Student (Id, Name, Branch) VALUES (%s, %s, %s)"
        val = (Id, Name, Branch)
        cursorobject.execute(sql, val)
        database.commit()
        print("Data inserted successfully!!!")
        
    elif choice == 3:
        query = "SELECT Id, Name, Branch FROM Student"
        cursorobject.execute(query)
        myresult = cursorobject.fetchall()
        for x in myresult:
            print(x)
            
    elif choice == 4:
        query = "UPDATE Student SET Name='hetu' WHERE Id=4"
        cursorobject.execute(query)
        database.commit() 
        print("Data Updated Successfully!!!")
        
    elif choice == 5:
        query = "DELETE FROM Student WHERE Id=4"
        cursorobject.execute(query)
        database.commit()
        print("Data Is Deleted Sucessfully!!!")
        
    elif choice == 6:
        print("Program ended.")
        break  
        
    else:
        print("Invalid choice! Please enter 1 to 6.")

cursorobject.close()
database.close()
```
