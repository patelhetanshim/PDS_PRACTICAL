```python
import math
def circle(radius):
    area=math.pi*radius*radius
    return area
radius=float(input("enter the radius of circle:"))
area=circle(radius)
print("Area of circle:",area)
```

```python
import statistics
def smodule(data):
    print("\n ststistics answer")
    print("data:",data)
    print("mean:",statistics.mean(data))
    print("median:",statistics.median(data))
    print("mode:",statistics.mode(data))
    print("stdev:",statistics.stdev(data))
    print("variance:",statistics.variance(data))
n=int(input("enter the number:"))
data=[]
for i in range(n):
  value=float(input("enter the value:"))
  data.append(value)
smodule(data)
```

```python
import random

data = input("Enter items separated by space: ").split()
count = int(input("How many unique items to select? "))

if count <= len(data):
    print("Selected Items:", random.sample(data, count))
else:
    print("Count cannot be greater than number of items.")
```
