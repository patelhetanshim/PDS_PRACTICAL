```python
import pandas as pd
import numpy as np

# ==========================================
# DATA PREPROCESSING USING PANDAS
# ==========================================

# Read CSV file
df = pd.read_csv("student details.csv")

print("Original Data:")
print(df.head())
```

```python
print("\n1. SLICING")

# First 5 rows
print(df.iloc[0:5])
```

```python
print("\n2. FILTERING")

# Students whose Marks are greater than 80
filtered = df[df["Marks"] > 80]

print(filtered[["Student Name", "Department", "Marks"]])
```

```python
print("\n3. SORTING")

# Sort by Marks in descending order
sorted_data = df.sort_values(
    "Marks",
    ascending=False
)

print(
    sorted_data[
        ["Student Name", "Department", "Marks"]
    ]
)
```

```python
print("\n4. CONCATENATING")

# Take first 2 rows and next 2 rows
data1 = df.iloc[0:2]
data2 = df.iloc[2:4]

combined = pd.concat(
    [data1, data2]
)

print(combined)
```

```python
print("\n5. AGGREGATION")

print("Total Marks:",
      df["Marks"].sum())

print("Average Marks:",
      df["Marks"].mean())

print("Maximum Marks:",
      df["Marks"].max())

print("Minimum Marks:",
      df["Marks"].min())
```

```python
print("\n6. NORMALIZATION")

# Normalize Marks between 0 and 1

minimum = df["Marks"].min()
maximum = df["Marks"].max()

df["NormalizedMarks"] = (
    (df["Marks"] - minimum)
    / (maximum - minimum)
)

print(
    df[
        ["Marks", "NormalizedMarks"]
    ]
)
```

```python
print("\nPreprocessing Completed!")

print("\nFinal Dataset:")
print(df)
```
