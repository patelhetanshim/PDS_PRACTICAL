```python
import numpy as np
import matplotlib.pyplot as plt

# ============================================================
# DYNAMIC MATPLOTLIB PLOTTING PROGRAM
# ============================================================

print("==============================================")
print("       MATPLOTLIB VISUALIZATION")
print("==============================================")

# ============================================================
# 1. LINE PLOT DATA
# ============================================================

print("\n===== LINE PLOT =====")

x_input = input(
    "Enter X values separated by spaces: "
)

y_input = input(
    "Enter Y values separated by spaces: "
)

x_line = np.array(
    [float(x) for x in x_input.split()]
)

y_line = np.array(
    [float(y) for y in y_input.split()]
)

if len(x_line) != len(y_line):
    raise ValueError(
        "X and Y must contain the same number of values."
    )


# ============================================================
# 2. BAR CHART DATA
# ============================================================

print("\n===== BAR CHART =====")

categories = input(
    "Enter categories separated by commas: "
).split(",")

bar_values_input = input(
    "Enter values separated by spaces: "
)

bar_values = np.array(
    [float(x) for x in bar_values_input.split()]
)

if len(categories) != len(bar_values):
    raise ValueError(
        "Number of categories and values must match."
    )

categories = [x.strip() for x in categories]


# ============================================================
# 3. HISTOGRAM DATA
# ============================================================

print("\n===== HISTOGRAM =====")

hist_input = input(
    "Enter numerical values separated by spaces: "
)

hist_data = np.array(
    [float(x) for x in hist_input.split()]
)

bins = int(
    input("Enter number of histogram bins: ")
)

if bins <= 0:
    raise ValueError(
        "Number of bins must be greater than 0."
    )


# ============================================================
# 4. PIE CHART DATA
# ============================================================

print("\n===== PIE CHART =====")

pie_labels = input(
    "Enter pie labels separated by commas: "
).split(",")

pie_values_input = input(
    "Enter pie values separated by spaces: "
)

pie_values = np.array(
    [float(x) for x in pie_values_input.split()]
)

if len(pie_labels) != len(pie_values):
    raise ValueError(
        "Number of labels and values must match."
    )

pie_labels = [x.strip() for x in pie_labels]


# ============================================================
# 5. SCATTER PLOT DATA
# ============================================================

print("\n===== SCATTER PLOT =====")

scatter_x_input = input(
    "Enter Scatter X values separated by spaces: "
)

scatter_y_input = input(
    "Enter Scatter Y values separated by spaces: "
)

scatter_x = np.array(
    [float(x) for x in scatter_x_input.split()]
)

scatter_y = np.array(
    [float(x) for x in scatter_y_input.split()]
)

if len(scatter_x) != len(scatter_y):
    raise ValueError(
        "Scatter X and Y must contain the same number of values."
    )


# ============================================================
# FIGURE AND SUBPLOTS
# ============================================================

fig, axes = plt.subplots(
    3,
    2,
    figsize=(14, 12)
)

# Overall figure title
fig.suptitle(
    "Matplotlib Data Visualization",
    fontsize=20,
    fontweight="bold"
)


# ============================================================
# LINE PLOT
# ============================================================

axes[0, 0].plot(
    x_line,
    y_line,
    marker="o",
    linewidth=2
)

axes[0, 0].set_title(
    "Line Plot"
)

axes[0, 0].set_xlabel(
    "X Values"
)

axes[0, 0].set_ylabel(
    "Y Values"
)

axes[0, 0].grid(
    True,
    alpha=0.3
)


# ============================================================
# BAR CHART
# ============================================================

axes[0, 1].bar(
    categories,
    bar_values
)

axes[0, 1].set_title(
    "Bar Chart"
)

axes[0, 1].set_xlabel(
    "Categories"
)

axes[0, 1].set_ylabel(
    "Values"
)

axes[0, 1].tick_params(
    axis="x",
    rotation=30
)


# ============================================================
# HISTOGRAM
# ============================================================

axes[1, 0].hist(
    hist_data,
    bins=bins,
    edgecolor="black",
    alpha=0.7
)

axes[1, 0].set_title(
    "Histogram"
)

axes[1, 0].set_xlabel(
    "Values"
)

axes[1, 0].set_ylabel(
    "Frequency"
)

axes[1, 0].grid(
    True,
    alpha=0.3
)


# ============================================================
# PIE CHART
# ============================================================

axes[1, 1].pie(
    pie_values,
    labels=pie_labels,
    autopct="%1.1f%%",
    startangle=90
)

axes[1, 1].set_title(
    "Pie Chart"
)


# ============================================================
# SCATTER PLOT
# ============================================================

axes[2, 0].scatter(
    scatter_x,
    scatter_y,
    s=70,
    alpha=0.7
)

axes[2, 0].set_title(
    "Scatter Plot"
)

axes[2, 0].set_xlabel(
    "X Values"
)

axes[2, 0].set_ylabel(
    "Y Values"
)

axes[2, 0].grid(
    True,
    alpha=0.3
)


# ============================================================
# EXTRA INFORMATION PLOT
# ============================================================

axes[2, 1].axis("off")

axes[2, 1].text(
    0.1,
    0.8,
    "Figure Properties",
    fontsize=15,
    fontweight="bold"
)

axes[2, 1].text(
    0.1,
    0.65,
    f"Figure Size: 14 × 12",
    fontsize=12
)

axes[2, 1].text(
    0.1,
    0.55,
    f"Line Points: {len(x_line)}",
    fontsize=12
)

axes[2, 1].text(
    0.1,
    0.45,
    f"Bar Categories: {len(categories)}",
    fontsize=12
)

axes[2, 1].text(
    0.1,
    0.35,
    f"Histogram Values: {len(hist_data)}",
    fontsize=12
)

axes[2, 1].text(
    0.1,
    0.25,
    f"Scatter Points: {len(scatter_x)}",
    fontsize=12
)


# ============================================================
# FIGURE PROPERTIES
# ============================================================

plt.tight_layout(
    rect=[0, 0, 1, 0.95]
)

plt.show()
```
