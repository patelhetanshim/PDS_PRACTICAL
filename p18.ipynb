```python
import pandas as pd
import numpy as np
# ============================================================
# DYNAMIC DATA CLEANING PROGRAM
# ============================================================
print("========================================")
print("        DATA CLEANING PROGRAM")
print("========================================")
# ------------------------------------------------------------
# Load dataset
# ------------------------------------------------------------
file_name = input("Enter CSV file name/path: ")
df = pd.read_csv(file_name)
print("\nOriginal Dataset:")
print(df)

print("\nDataset Shape:", df.shape)


# ============================================================
# 1. DETECT MISSING VALUES
# ============================================================

print("\n========================================")
print("        MISSING VALUES")
print("========================================")

missing_values = df.isnull().sum()

print("\nMissing values in each column:")
print(missing_values)

total_missing = missing_values.sum()

print("\nTotal missing values:", total_missing)


# ============================================================
# HANDLE MISSING VALUES
# ============================================================

if total_missing > 0:

    print("\nHow do you want to handle missing values?")
    print("1. Remove rows containing missing values")
    print("2. Fill numerical values with mean")
    print("3. Fill numerical values with median")
    print("4. Fill missing values with 0")

    choice = int(input("Enter choice (1/2/3/4): "))

    if choice == 1:

        df = df.dropna()

        print("\nRows containing missing values removed.")

    elif choice == 2:

        numeric_columns = df.select_dtypes(
            include=np.number
        ).columns

        for column in numeric_columns:
            df[column] = df[column].fillna(
                df[column].mean()
            )

        print("\nMissing numerical values filled with mean.")

    elif choice == 3:

        numeric_columns = df.select_dtypes(
            include=np.number
        ).columns

        for column in numeric_columns:
            df[column] = df[column].fillna(
                df[column].median()
            )

        print("\nMissing numerical values filled with median.")

    elif choice == 4:

        df = df.fillna(0)

        print("\nMissing values filled with 0.")

    else:
        raise ValueError("Invalid choice.")

else:

    print("\nNo missing values found.")


# ============================================================
# 2. REMOVE DUPLICATE RECORDS
# ============================================================

print("\n========================================")
print("        DUPLICATE RECORDS")
print("========================================")

duplicates = df.duplicated().sum()

print("Duplicate records found:", duplicates)

if duplicates > 0:

    remove_duplicates = input(
        "Remove duplicate records? (yes/no): "
    ).lower()

    if remove_duplicates == "yes":

        df = df.drop_duplicates()

        print("Duplicate records removed.")

    else:

        print("Duplicate records retained.")

else:

    print("No duplicate records found.")


# ============================================================
# 3. DETECT OUTLIERS USING IQR
# ============================================================

print("\n========================================")
print("        OUTLIER DETECTION")
print("========================================")

numeric_columns = df.select_dtypes(
    include=np.number
).columns

outlier_summary = {}

for column in numeric_columns:

    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df[
        (df[column] < lower_bound) |
        (df[column] > upper_bound)
    ]

    outlier_summary[column] = len(outliers)

    print(
        f"{column}: {len(outliers)} outliers"
    )


# ============================================================
# HANDLE OUTLIERS
# ============================================================

print("\nHow do you want to handle outliers?")
print("1. Remove outlier rows")
print("2. Replace outliers with median")
print("3. Keep outliers")

outlier_choice = int(
    input("Enter choice (1/2/3): ")
)


if outlier_choice == 1:

    for column in numeric_columns:

        Q1 = df[column].quantile(0.25)
        Q3 = df[column].quantile(0.75)

        IQR = Q3 - Q1

        lower_bound = Q1 - 1.5 * IQR
        upper_bound = Q3 + 1.5 * IQR

        df = df[
            (df[column] >= lower_bound) &
            (df[column] <= upper_bound)
        ]

    print("\nOutlier rows removed.")


elif outlier_choice == 2:

    for column in numeric_columns:

        Q1 = df[column].quantile(0.25)
        Q3 = df[column].quantile(0.75)

        IQR = Q3 - Q1

        lower_bound = Q1 - 1.5 * IQR
        upper_bound = Q3 + 1.5 * IQR

        median = df[column].median()

        df.loc[
            (df[column] < lower_bound) |
            (df[column] > upper_bound),
            column
        ] = median

    print("\nOutliers replaced with median.")


elif outlier_choice == 3:

    print("\nOutliers retained.")

else:

    raise ValueError("Invalid choice.")


# ============================================================
# FINAL DATASET
# ============================================================

print("\n========================================")
print("        CLEANED DATASET")
print("========================================")

print(df)

print("\nFinal Dataset Shape:", df.shape)

print("\nRemaining Missing Values:")
print(df.isnull().sum())

print("\nRemaining Duplicate Records:")
print(df.duplicated().sum())


# ============================================================
# SAVE CLEANED DATASET
# ============================================================

save_file = input(
    "\nEnter output CSV file name: "
)

df.to_csv(
    save_file,
    index=False
)

print(
    f"\nCleaned dataset saved as: {save_file}"
)

```
