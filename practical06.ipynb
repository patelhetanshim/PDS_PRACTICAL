```python
##Demonstrate exception handling using try, except, else, finally blocks and raise statement with custom exceptions.
try:
    x = 10
    y = x / 10
    print(x)
except NameError:                       # runs only if a variable is undefined
    print("Variable x is not defined")
except:                                 # runs for any other error
    print("Something went wrong")
```

```python
try:
    print("Hello")
except:
    print("Something went wrong")
else:
    print("Nothing went wrong")
```

```python
try:
    print(x)
except:
    print("Something went wrong")
else:
    print("else")
finally:
    print("always Execute")
```
